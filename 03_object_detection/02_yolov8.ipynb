````python
from ultralytics import YOLO
import os
import time

current_dir = os.getcwd()
print(f"현재 위치: {current_dir}")

# 1. 모델 불러오기 (가장 가볍고 빠른 nano 모델인 yolo8n.pt를 주로 쓴다.)
model = YOLO('yolov8n.pt')

start_time = time.time()

# 2. 모델 학습 시작하기
results = model.train(
    data='images/augmented_trashes50/data.yaml', 
    epochs=5, 
    imgsz=640, # YOLOv8의 기본 이미지 리사이즈 크기.
    device='mps', 

    project=os.path.join(current_dir, 'runs'),
    name='trash_v1'
)

end_time = time.time()

elapsed_time = end_time - start_time

mins = int(elapsed_time // 60)
secs = int(elapsed_time % 60)

print(f"소요시간: {mins}m {secs}s")
````
